```python
def plot_xro_stacked_onepanel(
    ds,
    region=1,
    start_date=None,
    end_date=None,
    figsize=(12, 8),
    line_alpha=0.9,
):
    da = ds.state_vector.sel(region=region)

    xro_states = [
        s for s in da.state.values
        if s != "PP"
    ]

    xro = da.sel(state=xro_states)

    if start_date is not None:
        xro = xro.sel(time=slice(start_date, None))
    if end_date is not None:
        xro = xro.sel(time=slice(None, end_date))

    time = xro.time.values
    values = xro.values  # [time, state]

    n_var = values.shape[1]

    fig, ax = plt.subplots(figsize=figsize)

    offset_step = 4.0
    offsets = np.arange(n_var)[::-1] * offset_step

    for i, state_name in enumerate(xro_states):
        y = values[:, i] + offsets[i]
        ax.plot(time, y, lw=1.2, alpha=line_alpha)
        ax.text(
            time[0],
            offsets[i],
            state_name,
            va="center",
            ha="right",
            fontsize=10,
        )

        ax.axhline(
            offsets[i],
            color="0.85",
            lw=0.8,
            zorder=0,
        )

    ax.set_title(f"Region {region}: XRO climate modes")
    ax.set_xlabel("Time")
    ax.set_yticks([])
    ax.grid(axis="x", alpha=0.2)
    plt.tight_layout()
    plt.show()
```

```python
plot_xro_stacked_onepanel(
    ds,
    region=1,
    start_date="1998-01-01",
    end_date="2025-12-31",
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

EXCLUDED_REGIONS = {
    6,
    18,
    19,
    29,
}

def _as_string(value):
    """Convert byte strings or scalar arrays to a regular Python string."""
    if hasattr(value, "item"):
        value = value.item()

    if isinstance(value, bytes):
        return value.decode("utf-8")

    return str(value)


def _infer_ocean_group(
    basin_name,
    region_name,
):
    """
    Assign each region to a broad ocean-basin group.

    Returns one of:
        ATL, MED, IO, PO, AO, SO
    """
    basin = _as_string(
        basin_name
    ).strip().lower()

    region = _as_string(
        region_name
    ).strip().upper()

    # Treat the Mediterranean as a separate group.
    if region.startswith("MED"):
        return "MED"

    if basin == "atlantic":
        return "ATL"

    if basin == "indian":
        return "IO"

    if basin == "pacific":
        return "PO"

    if basin == "arctic":
        return "AO"

    if basin == "southern":
        return "SO"

    return None


def plot_pp_stacked_by_ocean(
    ds,
    start_date=None,
    end_date=None,
    running_mean_window=1,
    offset_step=4.0,
    figsize=(14, 18),
    line_width=1.0,
    line_alpha=0.9,
):
    """
    Plot standardized regional PP time series as stacked lines,
    grouped into five ocean panels.

    Parameters
    ----------
    ds : xarray.Dataset
        Dataset containing state_vector(region, time, state).

    start_date, end_date : str or None
        Optional plotting period.

    running_mean_window : int
        Set to 1 for monthly values or 3 for a centered
        three-month running mean.

    offset_step : float
        Vertical spacing between regional time series.
    """
    pp = ds[
        "state_vector"
    ].sel(
        state="PP"
    )

    if start_date is not None:
        pp = pp.sel(
            time=slice(
                start_date,
                None,
            )
        )

    if end_date is not None:
        pp = pp.sel(
            time=slice(
                None,
                end_date,
            )
        )

    if running_mean_window > 1:
        pp = pp.rolling(
            time=running_mean_window,
            center=True,
            min_periods=running_mean_window,
        ).mean()

    regions = pp[
        "region"
    ].values

    # Exclude selected regions from the plot.
    regions = [
        region
        for region in regions
        if int(region) not in EXCLUDED_REGIONS
    ]

    ocean_groups = {
        "ATL": [],
        "MED": [],
        "IO": [],
        "PO": [],
        "AO": [],
        "SO": [],
    }

    unmatched_regions = []

    for region in regions:
        if "basin" in ds:
            basin_name = ds[
                "basin"
            ].sel(
                region=region
            ).values
        else:
            basin_name = ""

        if "region_name" in ds:
            region_name = ds[
                "region_name"
            ].sel(
                region=region
            ).values
        else:
            region_name = (
                f"Region {int(region)}"
            )

        ocean_group = _infer_ocean_group(
            basin_name=basin_name,
            region_name=region_name,
        )

        if ocean_group is None:
            unmatched_regions.append(
                {
                    "region": int(region),
                    "basin": _as_string(
                        basin_name
                    ),
                    "region_name": _as_string(
                        region_name
                    ),
                }
            )

            continue

        ocean_groups[
            ocean_group
        ].append(
            region
        )

    if unmatched_regions:
        print(
            "Regions not assigned to an ocean group:"
        )

        for item in unmatched_regions:
            print(
                item
            )

    ocean_order = [
        "ATL",
        "MED",
        "IO",
        "PO",
        "AO",
        "SO",
    ]

    ocean_titles = {
        "ATL": "Atlantic Ocean",
        "MED": "Mediterranean Sea",
        "IO": "Indian Ocean",
        "PO": "Pacific Ocean",
        "AO": "Arctic Ocean",
        "SO": "Southern Ocean",
    }

    # Remove empty groups while preserving the requested ocean order.
    ocean_order = [
        ocean
        for ocean in ocean_order
        if len(
            ocean_groups[
                ocean
            ]
        ) > 0
    ]

    # Give panels with more regions proportionally more vertical space.
    height_ratios = [
        len(
            ocean_groups[
                ocean
            ]
        )
        for ocean in ocean_order
    ]

    fig, axes = plt.subplots(
        nrows=len(
            ocean_order
        ),
        ncols=1,
        figsize=figsize,
        sharex=True,
        gridspec_kw={
            "height_ratios": (
                height_ratios
            ),
            "hspace": 0.12,
        },
    )

    if len(
        ocean_order
    ) == 1:
        axes = [
            axes
        ]

    time = pp[
        "time"
    ].values

    for ax, ocean in zip(
        axes,
        ocean_order,
    ):
        ocean_regions = sorted(
            ocean_groups[
                ocean
            ],
            key=lambda value: int(
                value
            ),
        )

        n_regions = len(
            ocean_regions
        )

        offsets = (
            np.arange(
                n_regions
            )[::-1]
            * offset_step
        )

        ytick_labels = []

        for region_index, region in enumerate(
            ocean_regions
        ):
            offset = offsets[
                region_index
            ]

            values = pp.sel(
                region=region
            ).values

            shifted_values = (
                values
                + offset
            )

            ax.plot(
                time,
                shifted_values,
                linewidth=line_width,
                alpha=line_alpha,
            )

            # Mark the zero-anomaly level for each regional series.
            ax.axhline(
                offset,
                linewidth=0.7,
                color="0.85",
                zorder=0,
            )

            if "region_name" in ds:
                region_name = _as_string(
                    ds[
                        "region_name"
                    ].sel(
                        region=region
                    ).values
                )
            else:
                region_name = (
                    f"Region {int(region)}"
                )

            ytick_labels.append(
                f"R{int(region)}  "
                f"{region_name}"
            )

        ax.set_yticks(
            offsets
        )

        ax.set_yticklabels(
            ytick_labels,
            fontsize=9,
        )

        ax.set_title(
            ocean_titles[
                ocean
            ],
            loc="left",
            fontsize=12,
            fontweight="bold",
        )

        ax.grid(
            axis="x",
            alpha=0.2,
        )

        ax.tick_params(
            axis="y",
            length=0,
        )

        ax.spines[
            "top"
        ].set_visible(
            False
        )

        ax.spines[
            "right"
        ].set_visible(
            False
        )

        ax.spines[
            "left"
        ].set_visible(
            False
        )

    axes[
        -1
    ].set_xlabel(
        "Time"
    )

    smoothing_text = (
        "monthly"
        if running_mean_window == 1
        else (
            f"{running_mean_window}-month "
            "centered mean"
        )
    )

    fig.suptitle(
        "Regional standardized PP anomalies "
        f"({smoothing_text})",
        fontsize=15,
        y=0.995,
    )

    plt.subplots_adjust(
        left=0.27,
        right=0.98,
        top=0.97,
        bottom=0.05,
    )

    plt.show()
```

```python
plot_pp_stacked_by_ocean(
    ds,
    start_date="1998-01-01",
    end_date="2025-12-31",
    running_mean_window=1,
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import math


def plot_state_distributions(
    ds,
    region=1,
    start_date=None,
    end_date=None,
    include_pp=True,
    ncols=4,
    bins="fd",
    figsize=(13.33, 7.5),   # 16:9 PowerPoint
):
    """
    Plot compact distributions of preprocessed state variables.

    Parameters
    ----------
    ds : xr.Dataset
        Must contain ds.state_vector(time, state, region).
    region : int
        Region number.
    start_date, end_date : str or None
        Optional time range.
    include_pp : bool
        Include PP in addition to XRO climate-mode inputs.
    ncols : int
        Number of subplot columns.
    bins : int or str
        Histogram bin specification.
    figsize : tuple
        Figure size. Default corresponds approximately to 16:9 PPT.
    """

    da = ds.state_vector.sel(region=region)

    state_names = [str(s) for s in da.state.values]

    if not include_pp:
        state_names = [
            s for s in state_names
            if s != "PP"
        ]

    da = da.sel(state=state_names)

    if start_date is not None:
        da = da.sel(time=slice(start_date, None))

    if end_date is not None:
        da = da.sel(time=slice(None, end_date))

    da = da.transpose("time", "state")

    n_var = len(state_names)
    nrows = math.ceil(n_var / ncols)

    fig, axes = plt.subplots(
        nrows,
        ncols,
        figsize=figsize,
        squeeze=False,
    )

    axes = axes.ravel()

    for i, state_name in enumerate(state_names):

        ax = axes[i]

        values = da.sel(state=state_name).values
        values = values[np.isfinite(values)]

        if values.size < 8:
            ax.text(
                0.5,
                0.5,
                "Insufficient data",
                ha="center",
                va="center",
                transform=ax.transAxes,
            )
            ax.set_title(state_name)
            continue

        # Histogram
        ax.hist(
            values,
            bins=bins,
            density=True,
            alpha=0.4,
            label="Histogram",
        )

        sigma = np.std(values, ddof=1)

        if sigma > 0:

            x_grid = np.linspace(
                np.min(values),
                np.max(values),
                400,
            )

            # Empirical KDE
            try:
                kde = stats.gaussian_kde(values)

                ax.plot(
                    x_grid,
                    kde(x_grid),
                    lw=1.7,
                    label="KDE",
                )
            except np.linalg.LinAlgError:
                pass

            # Gaussian with the same mean and standard deviation
            mu = np.mean(values)

            ax.plot(
                x_grid,
                stats.norm.pdf(
                    x_grid,
                    loc=mu,
                    scale=sigma,
                ),
                ls="--",
                lw=1.5,
                label="Gaussian",
            )

        skew = stats.skew(
            values,
            bias=False,
        )

        kurt = stats.kurtosis(
            values,
            fisher=True,
            bias=False,
        )

        p95 = np.percentile(values, 95)
        p99 = np.percentile(values, 99)

        ax.set_title(
            f"{state_name}\n"
            f"skew={skew:.2f}, kurt={kurt:.2f}",
            fontsize=10,
        )

        ax.text(
            0.98,
            0.96,
            f"P95={p95:.2f}\nP99={p99:.2f}",
            transform=ax.transAxes,
            ha="right",
            va="top",
            fontsize=7.5,
        )

        ax.set_xlabel("Preprocessed value", fontsize=8)
        ax.set_ylabel("Density", fontsize=8)
        ax.tick_params(labelsize=7.5)
        ax.grid(alpha=0.2)

    # Remove empty axes
    for j in range(n_var, len(axes)):
        axes[j].axis("off")

    # One common legend instead of a legend in every panel
    handles, labels = axes[0].get_legend_handles_labels()

    if handles:
        fig.legend(
            handles,
            labels,
            loc="upper center",
            bbox_to_anchor=(0.5, 0.955),
            ncol=3,
            frameon=False,
            fontsize=9,
        )

    fig.suptitle(
        f"Region {region}: distributions of preprocessed state variables",
        fontsize=14,
        y=0.995,
    )

    fig.subplots_adjust(
        left=0.06,
        right=0.98,
        bottom=0.08,
        top=0.88,
        wspace=0.30,
        hspace=0.48,
    )

    plt.show()
```

```python
plot_state_distributions(
    ds,
    region=1,
    start_date="1998-01-01",
    end_date="2025-12-31",
    include_pp=True,
    ncols=4,
)
```

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats


def plot_pp_distributions_by_region(
    ds,
    regions=None,
    exclude_regions=(6, 18, 19, 29),
    start_date=None,
    end_date=None,
    region_labels=None,
    ncols=6,
    bins="fd",
    figsize=(13.33, 7.5),   # 16:9 PowerPoint
    common_xlim=False,
    savepath=None,
):
    """
    Plot preprocessed PP distributions separately by region.

    R6, R18, R19, and R29 are excluded by default.
    """

    pp = ds.state_vector.sel(state="PP")

    if start_date is not None:
        pp = pp.sel(time=slice(start_date, None))

    if end_date is not None:
        pp = pp.sel(time=slice(None, end_date))

    available_regions = list(pp.region.values)

    if regions is None:
        regions = available_regions
    else:
        regions = [
            r for r in regions
            if r in available_regions
        ]

    # Exclude selected regions
    exclude_regions = set(exclude_regions)

    regions = [
        r for r in regions
        if int(r) not in exclude_regions
    ]

    pp = pp.sel(region=regions)

    nregions = len(regions)
    nrows = math.ceil(nregions / ncols)

    fig, axes = plt.subplots(
        nrows,
        ncols,
        figsize=figsize,
        squeeze=False,
    )

    axes = axes.ravel()

    # Optional common x-axis range
    if common_xlim:
        all_values = pp.values.ravel()
        all_values = all_values[np.isfinite(all_values)]

        xmin = np.min(all_values)
        xmax = np.max(all_values)
    else:
        xmin = xmax = None

    legend_handles = None
    legend_labels = None

    for i, region in enumerate(regions):

        ax = axes[i]

        values = pp.sel(region=region).values
        values = values[np.isfinite(values)]

        if region_labels is None:
            label = f"R{int(region)}"
        else:
            label = region_labels.get(
                int(region),
                f"R{int(region)}",
            )

        if values.size < 8:
            ax.text(
                0.5,
                0.5,
                "Insufficient data",
                ha="center",
                va="center",
                transform=ax.transAxes,
            )
            ax.set_title(label, fontsize=9)
            continue

        # Histogram
        ax.hist(
            values,
            bins=bins,
            density=True,
            alpha=0.4,
            label="Histogram",
        )

        sigma = np.std(values, ddof=1)

        if sigma > 0 and np.unique(values).size > 1:

            x_grid = np.linspace(
                np.min(values),
                np.max(values),
                300,
            )

            # Empirical KDE
            try:
                kde = stats.gaussian_kde(values)

                ax.plot(
                    x_grid,
                    kde(x_grid),
                    lw=1.4,
                    label="KDE",
                )
            except np.linalg.LinAlgError:
                pass

            # Gaussian with the same mean and standard deviation
            mu = np.mean(values)

            ax.plot(
                x_grid,
                stats.norm.pdf(
                    x_grid,
                    loc=mu,
                    scale=sigma,
                ),
                ls="--",
                lw=1.2,
                label="Gaussian",
            )

        skew = stats.skew(
            values,
            bias=False,
        )

        kurt = stats.kurtosis(
            values,
            fisher=True,
            bias=False,
        )

        p99 = np.percentile(values, 99)
        vmax = np.max(values)

        ax.set_title(
            f"{label}\n"
            f"skew={skew:.2f}, kurt={kurt:.2f}",
            fontsize=8.5,
        )

        ax.text(
            0.97,
            0.94,
            f"P99={p99:.2f}\nmax={vmax:.2f}",
            transform=ax.transAxes,
            ha="right",
            va="top",
            fontsize=6.5,
        )

        if common_xlim:
            ax.set_xlim(xmin, xmax)

        ax.tick_params(labelsize=6.5)
        ax.grid(alpha=0.15)

        if i % ncols == 0:
            ax.set_ylabel("Density", fontsize=7)

        if i >= (nrows - 1) * ncols:
            ax.set_xlabel("PP anomaly", fontsize=7)

        if legend_handles is None:
            legend_handles, legend_labels = (
                ax.get_legend_handles_labels()
            )

    # Turn off unused panels
    for j in range(nregions, len(axes)):
        axes[j].axis("off")

    if legend_handles:
        fig.legend(
            legend_handles,
            legend_labels,
            loc="upper center",
            bbox_to_anchor=(0.5, 0.955),
            ncol=3,
            frameon=False,
            fontsize=8,
        )

    fig.suptitle(
        "Regional distributions of preprocessed PP",
        fontsize=13,
        y=0.995,
    )

    fig.subplots_adjust(
        left=0.045,
        right=0.985,
        bottom=0.065,
        top=0.88,
        wspace=0.30,
        hspace=0.52,
    )

    if savepath is not None:
        fig.savefig(
            savepath,
            dpi=300,
            bbox_inches="tight",
        )

    plt.show()
```

```python
plot_pp_distributions_by_region(
    ds,
    start_date="1998-01-01",
    end_date="2025-12-31",
)
```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt

from matplotlib.colors import BoundaryNorm
from matplotlib.transforms import blended_transform_factory
from scipy.stats import pearsonr


# ============================================================
# Helpers
# ============================================================

def _as_text(value):
    if isinstance(value, bytes):
        return value.decode("utf-8")
    return str(value)


def _find_region_field(ds, candidates):
    """
    Find a region-dependent coordinate or data variable.
    """

    for name in candidates:

        if name in ds.coords and "region" in ds.coords[name].dims:
            return ds.coords[name]

        if name in ds.data_vars and "region" in ds[name].dims:
            return ds[name]

        if (
            hasattr(ds, "state_vector")
            and name in ds.state_vector.coords
            and "region" in ds.state_vector.coords[name].dims
        ):
            return ds.state_vector.coords[name]

    return None


def _normalise_ocean_name(value):
    """
    Convert basin/ocean names to:
    SO, IO, PO, AT, Other
    """

    text = _as_text(value).strip().upper()

    if text in {"SO", "SOUTHERN", "SOUTHERN OCEAN"}:
        return "SO"

    if text in {"IO", "INDIAN", "INDIAN OCEAN"}:
        return "IO"

    if text in {"PO", "PACIFIC", "PACIFIC OCEAN"}:
        return "PO"

    # All Atlantic regions grouped together
    if (
        text in {
            "AT",
            "ATLANTIC",
            "ATLANTIC OCEAN",
            "NA",
            "NORTH ATLANTIC",
            "NORTH ATLANTIC OCEAN",
            "SA",
            "SOUTH ATLANTIC",
            "SOUTH ATLANTIC OCEAN",
            "AE",
            "AEQU",
            "EQUATORIAL ATLANTIC",
            "ATLANTIC EQUATORIAL",
        }
        or "ATLANTIC" in text
    ):
        return "Atlantic Ocean"

    return "Other"


def _infer_ocean_from_region_name(region_name):
    """
    Infer ocean group from region names.

    NA, SA, and AEQU are all grouped as AT.
    """

    text = _as_text(region_name).strip().upper()

    # Atlantic region names
    if "AEQU" in text:
        return "Atlantic Ocean"

    tokens = (
        text.replace("-", " ")
        .replace("_", " ")
        .split()
    )

    # NA and SA must be checked before generic prefixes
    if "NA" in tokens or "SA" in tokens or "AE" in tokens:
        return "Atlantic Ocean"

    if "ATLANTIC" in tokens:
        return "Atlantic Ocean"

    if "SO" in tokens or text.startswith("SO"):
        return "SO"

    if "IO" in tokens or text.startswith("IO"):
        return "IO"

    if "PO" in tokens or text.startswith("PO"):
        return "PO"

    return "Other"


def get_region_metadata(ds):
    """
    Read region metadata directly from the dataset.

    Candidate fields
    ----------------
    region name:
        region_name, region_label, name

    ocean/basin:
        basin, ocean, ocean_group, basin_name

    original region ID:
        original_region_id, region_id, original_id
    """

    region_name_da = _find_region_field(
        ds,
        ["region_name", "region_label", "name"],
    )

    basin_da = _find_region_field(
        ds,
        ["basin", "ocean", "ocean_group", "basin_name"],
    )

    original_id_da = _find_region_field(
        ds,
        ["original_region_id", "region_id", "original_id"],
    )

    metadata = {}

    for region in ds.state_vector.region.values:

        region_key = int(region)

        # Original region ID
        if original_id_da is not None:
            try:
                original_id = int(
                    original_id_da.sel(region=region).item()
                )
            except (TypeError, ValueError):
                original_id = region_key
        else:
            original_id = region_key

        # Region name
        if region_name_da is not None:
            try:
                region_name = _as_text(
                    region_name_da.sel(region=region).item()
                ).strip()
            except Exception:
                region_name = f"R{original_id}"
        else:
            region_name = f"R{original_id}"

        # Ocean group
        ocean_from_name = _infer_ocean_from_region_name(
            region_name
        )

        if ocean_from_name != "Other":
            ocean = ocean_from_name

        elif basin_da is not None:
            try:
                ocean = _normalise_ocean_name(
                    basin_da.sel(region=region).item()
                )
            except Exception:
                ocean = "Other"

        else:
            ocean = "Other"

        # Avoid labels such as "R1 R1 NA SPSS"
        compact_name = region_name.strip()

        if compact_name.upper().startswith(
            f"R{original_id}"
        ):
            display_label = compact_name
        else:
            display_label = f"R{original_id}  {compact_name}"

        metadata[region_key] = {
            "original_region_id": original_id,
            "region_name": region_name,
            "display_label": display_label,
            "ocean": ocean,
        }

    return metadata


def _pairwise_corr(x, y, min_samples):
    """
    Pearson correlation with pairwise finite-value selection.
    """

    valid = np.isfinite(x) & np.isfinite(y)
    n = int(valid.sum())

    if n < min_samples:
        return np.nan, n

    xv = np.asarray(x[valid], dtype=float)
    yv = np.asarray(y[valid], dtype=float)

    xv = xv - xv.mean()
    yv = yv - yv.mean()

    denominator = np.sqrt(
        np.sum(xv ** 2) * np.sum(yv ** 2)
    )

    if denominator == 0:
        return np.nan, n

    correlation = np.sum(xv * yv) / denominator

    return float(correlation), n


def _max_abs_lag_correlation(
    x,
    y,
    max_lead,
    min_samples,
):
    """
    Find maximum absolute corr[X(t), Y(t + lead)].
    """

    best_corr = np.nan
    best_lead = -1
    best_n = 0

    for lead in range(max_lead + 1):

        if lead == 0:
            xx = x
            yy = y
        else:
            xx = x[:-lead]
            yy = y[lead:]

        r, n = _pairwise_corr(
            xx,
            yy,
            min_samples=min_samples,
        )

        if not np.isfinite(r):
            continue

        if (
            not np.isfinite(best_corr)
            or abs(r) > abs(best_corr)
        ):
            best_corr = r
            best_lead = lead
            best_n = n

    return best_corr, best_lead, best_n


def _circular_shift_maxlag_pvalue(
    x,
    y,
    observed_max_abs_corr,
    max_lead,
    min_samples,
    n_permutations,
    rng,
):
    """
    Circular-shift permutation test.

    The null distribution is based on the maximum absolute
    correlation across all leads, so the p-value accounts for
    searching multiple leads within each region-predictor cell.
    """

    n_time = len(x)

    minimum_shift = max_lead + 1
    maximum_shift = n_time - max_lead - 1

    if maximum_shift <= minimum_shift:
        shift_candidates = np.arange(1, n_time)
    else:
        shift_candidates = np.arange(
            minimum_shift,
            maximum_shift + 1,
        )

    if shift_candidates.size == 0:
        return np.nan

    replace = n_permutations > shift_candidates.size

    shifts = rng.choice(
        shift_candidates,
        size=n_permutations,
        replace=replace,
    )

    null_max_corr = np.full(
        n_permutations,
        np.nan,
    )

    for i, shift in enumerate(shifts):

        x_shifted = np.roll(x, int(shift))

        r_null, _, _ = _max_abs_lag_correlation(
            x=x_shifted,
            y=y,
            max_lead=max_lead,
            min_samples=min_samples,
        )

        if np.isfinite(r_null):
            null_max_corr[i] = abs(r_null)

    valid_null = null_max_corr[
        np.isfinite(null_max_corr)
    ]

    if valid_null.size == 0:
        return np.nan

    pvalue = (
        1
        + np.sum(
            valid_null >= observed_max_abs_corr
        )
    ) / (
        valid_null.size + 1
    )

    return float(pvalue)


def _benjamini_hochberg(pvalues):
    """
    Benjamini-Hochberg FDR-adjusted q-values.
    """

    pvalues = np.asarray(pvalues, dtype=float)
    qvalues = np.full_like(pvalues, np.nan)

    finite = np.isfinite(pvalues)

    if not np.any(finite):
        return qvalues

    p = pvalues[finite]
    order = np.argsort(p)
    ranked = p[order]

    n = ranked.size

    adjusted = ranked * n / np.arange(1, n + 1)

    # Enforce monotonicity
    adjusted = np.minimum.accumulate(
        adjusted[::-1]
    )[::-1]

    adjusted = np.clip(adjusted, 0, 1)

    restored = np.empty_like(adjusted)
    restored[order] = adjusted

    qvalues[finite] = restored

    return qvalues


# ============================================================
# Main function
# ============================================================

def plot_pp_xro_best_lag_heatmap(
    ds,
    regions=None,
    exclude_regions=(6, 18, 19, 29),
    predictors=None,
    start_date=None,
    end_date=None,
    max_lead=24,
    pp_rolling=1,
    rolling_center=False,
    min_samples=60,
    ocean_order=("SO", "IO", "PO", "NA", "SA", "Other"),
    ocean_display=None,
    colour_range=None,
    colour_step=0.1,
    auto_colour_percentile=95,
    minimum_colour_range=0.3,
    significance_test="permutation",
    star_rule="p",
    alpha=0.05,
    sig_r=0.4,
    n_permutations=500,
    apply_fdr=False,
    random_seed=42,
    show_lead_numbers=True,
    figsize=(13.33, 7.5),
    savepath=None,
):
    """
    Plot strongest PP-XRO lead-lag correlations.

    Correlation definition
    ----------------------
    corr[XRO(t), PP(t + lead)]

    lead > 0 means that XRO leads PP.

    Cell colour
    -----------
    Signed correlation at the lead having the largest
    absolute correlation.

    Cell number
    -----------
    XRO lead month.

    Star
    ----
    Controlled using star_rule:

        "p"      : p < alpha
        "corr"   : abs(r) >= sig_r
        "either" : p < alpha OR abs(r) >= sig_r
        "both"   : p < alpha AND abs(r) >= sig_r
        None     : no stars

    significance_test
    -----------------
        "permutation":
            Circular-shift max-lag permutation test.
            This accounts for selecting the strongest correlation
            among leads 0...max_lead.

        "pearson":
            Ordinary Pearson p-value at the selected lead.
            This does not correct for searching multiple leads.

        None:
            No p-value test.

    apply_fdr
    ---------
    If True, Benjamini-Hochberg correction is additionally
    applied across all region-predictor cells.

    Returns
    -------
    result : xr.Dataset
    """

    if "state_vector" not in ds:
        raise ValueError(
            "Dataset must contain 'state_vector'."
        )

    da = ds.state_vector

    required_dims = {"time", "state", "region"}

    if not required_dims.issubset(set(da.dims)):
        raise ValueError(
            "state_vector must have time, state, and region dimensions."
        )

    # ========================================================
    # Time selection
    # ========================================================
    if start_date is not None:
        da = da.sel(
            time=slice(start_date, None)
        )

    if end_date is not None:
        da = da.sel(
            time=slice(None, end_date)
        )

    # ========================================================
    # Identify states
    # ========================================================
    state_values = list(da.state.values)

    pp_matches = [
        state
        for state in state_values
        if str(state) == "PP"
    ]

    if len(pp_matches) != 1:
        raise ValueError(
            "Could not identify one unique PP state."
        )

    pp_state = pp_matches[0]

    state_lookup = {
        str(state): state
        for state in state_values
    }

    if predictors is None:
        predictor_values = [
            state
            for state in state_values
            if str(state) != "PP"
        ]
    else:
        missing = [
            predictor
            for predictor in predictors
            if str(predictor) not in state_lookup
        ]

        if missing:
            raise ValueError(
                f"Predictors not found: {missing}"
            )

        predictor_values = [
            state_lookup[str(predictor)]
            for predictor in predictors
        ]

    predictor_names = [
        str(predictor)
        for predictor in predictor_values
    ]

    # ========================================================
    # Region metadata and selection
    # ========================================================
    region_metadata = get_region_metadata(ds)

    available_regions = list(da.region.values)

    if regions is None:
        selected_regions = available_regions
    else:
        requested = {int(region) for region in regions}

        selected_regions = [
            region
            for region in available_regions
            if (
                int(region) in requested
                or region_metadata[int(region)][
                    "original_region_id"
                ] in requested
            )
        ]

    excluded = {
        int(region)
        for region in exclude_regions
    }

    selected_regions = [
        region
        for region in selected_regions
        if (
            int(region) not in excluded
            and region_metadata[int(region)][
                "original_region_id"
            ] not in excluded
        )
    ]

    ocean_rank = {
        ocean: i
        for i, ocean in enumerate(ocean_order)
    }

    selected_regions = sorted(
        selected_regions,
        key=lambda region: (
            ocean_rank.get(
                region_metadata[int(region)]["ocean"],
                999,
            ),
            region_metadata[int(region)][
                "original_region_id"
            ],
        ),
    )

    if len(selected_regions) == 0:
        raise ValueError(
            "No regions remain after exclusion."
        )

    # ========================================================
    # Extract PP and XRO
    # ========================================================
    pp = da.sel(
        state=pp_state,
        region=selected_regions,
    )

    xro = da.sel(
        state=predictor_values,
        region=selected_regions,
    )

    if pp_rolling > 1:
        pp = pp.rolling(
            time=pp_rolling,
            center=rolling_center,
            min_periods=pp_rolling,
        ).mean()

    n_region = len(selected_regions)
    n_predictor = len(predictor_values)

    leads = np.arange(max_lead + 1)

    corr = np.full(
        (n_region, n_predictor, len(leads)),
        np.nan,
    )

    sample_size = np.zeros(
        (n_region, n_predictor, len(leads)),
        dtype=int,
    )

    best_corr = np.full(
        (n_region, n_predictor),
        np.nan,
    )

    best_lead = np.full(
        (n_region, n_predictor),
        -1,
        dtype=int,
    )

    best_sample_size = np.zeros(
        (n_region, n_predictor),
        dtype=int,
    )

    pearson_p = np.full(
        (n_region, n_predictor),
        np.nan,
    )

    permutation_p = np.full(
        (n_region, n_predictor),
        np.nan,
    )

    rng = np.random.default_rng(random_seed)

    # ========================================================
    # Correlations and significance
    # ========================================================
    for ir, region in enumerate(selected_regions):

        y_full = np.asarray(
            pp.sel(region=region).values,
            dtype=float,
        )

        for ip, predictor in enumerate(predictor_values):

            x_full = np.asarray(
                xro.sel(
                    region=region,
                    state=predictor,
                ).values,
                dtype=float,
            )

            # All lead correlations
            for il, lead in enumerate(leads):

                if lead == 0:
                    xx = x_full
                    yy = y_full
                else:
                    xx = x_full[:-lead]
                    yy = y_full[lead:]

                r, n = _pairwise_corr(
                    xx,
                    yy,
                    min_samples=min_samples,
                )

                corr[ir, ip, il] = r
                sample_size[ir, ip, il] = n

            values = corr[ir, ip, :]

            if not np.any(np.isfinite(values)):
                continue

            best_index = int(
                np.nanargmax(np.abs(values))
            )

            selected_r = values[best_index]
            selected_lead = int(leads[best_index])

            best_corr[ir, ip] = selected_r
            best_lead[ir, ip] = selected_lead
            best_sample_size[ir, ip] = (
                sample_size[ir, ip, best_index]
            )

            # Ordinary Pearson p-value at selected lead
            if selected_lead == 0:
                xx = x_full
                yy = y_full
            else:
                xx = x_full[:-selected_lead]
                yy = y_full[selected_lead:]

            valid = np.isfinite(xx) & np.isfinite(yy)

            if valid.sum() >= min_samples:
                try:
                    _, p = pearsonr(
                        xx[valid],
                        yy[valid],
                    )
                    pearson_p[ir, ip] = p
                except Exception:
                    pass

            # Max-lag permutation p-value
            if significance_test == "permutation":

                permutation_p[ir, ip] = (
                    _circular_shift_maxlag_pvalue(
                        x=x_full,
                        y=y_full,
                        observed_max_abs_corr=abs(
                            selected_r
                        ),
                        max_lead=max_lead,
                        min_samples=min_samples,
                        n_permutations=n_permutations,
                        rng=rng,
                    )
                )

    # ========================================================
    # Select p-value field
    # ========================================================
    if significance_test == "permutation":
        selected_p = permutation_p.copy()

    elif significance_test == "pearson":
        selected_p = pearson_p.copy()

    elif significance_test is None:
        selected_p = np.full_like(
            best_corr,
            np.nan,
        )

    else:
        raise ValueError(
            "significance_test must be "
            "'permutation', 'pearson', or None."
        )

    if apply_fdr:
        selected_q = _benjamini_hochberg(
            selected_p
        )
        p_for_stars = selected_q
    else:
        selected_q = np.full_like(
            selected_p,
            np.nan,
        )
        p_for_stars = selected_p

    # ========================================================
    # Star mask
    # ========================================================
    p_condition = (
        np.isfinite(p_for_stars)
        & (p_for_stars < alpha)
    )

    r_condition = (
        np.isfinite(best_corr)
        & (np.abs(best_corr) >= sig_r)
    )

    if star_rule is None:
        significant = np.zeros_like(
            best_corr,
            dtype=bool,
        )

    elif star_rule == "p":
        significant = p_condition

    elif star_rule == "corr":
        significant = r_condition

    elif star_rule == "either":
        significant = p_condition | r_condition

    elif star_rule == "both":
        significant = p_condition & r_condition

    else:
        raise ValueError(
            "star_rule must be 'p', 'corr', "
            "'either', 'both', or None."
        )

    # ========================================================
    # Build output dataset
    # ========================================================
    region_labels = [
        region_metadata[int(region)]["display_label"]
        for region in selected_regions
    ]

    region_names = [
        region_metadata[int(region)]["region_name"]
        for region in selected_regions
    ]

    region_oceans = [
        region_metadata[int(region)]["ocean"]
        for region in selected_regions
    ]

    original_ids = [
        region_metadata[int(region)]["original_region_id"]
        for region in selected_regions
    ]

    result = xr.Dataset(
        data_vars={
            "corr": (
                ("region", "predictor", "lead"),
                corr,
            ),
            "sample_size": (
                ("region", "predictor", "lead"),
                sample_size,
            ),
            "best_corr": (
                ("region", "predictor"),
                best_corr,
            ),
            "best_lead": (
                ("region", "predictor"),
                best_lead,
            ),
            "best_sample_size": (
                ("region", "predictor"),
                best_sample_size,
            ),
            "pearson_p": (
                ("region", "predictor"),
                pearson_p,
            ),
            "permutation_p": (
                ("region", "predictor"),
                permutation_p,
            ),
            "selected_p": (
                ("region", "predictor"),
                selected_p,
            ),
            "selected_q": (
                ("region", "predictor"),
                selected_q,
            ),
            "significant": (
                ("region", "predictor"),
                significant,
            ),
        },
        coords={
            "region": selected_regions,
            "predictor": predictor_names,
            "lead": leads,
            "original_region_id": (
                "region",
                original_ids,
            ),
            "region_name": (
                "region",
                region_names,
            ),
            "region_label": (
                "region",
                region_labels,
            ),
            "ocean": (
                "region",
                region_oceans,
            ),
        },
    )

    result.attrs[
        "correlation_definition"
    ] = "corr[XRO(t), PP(t + lead)]"

    result.attrs[
        "significance_test"
    ] = str(significance_test)

    result.attrs[
        "apply_fdr"
    ] = bool(apply_fdr)

    # ========================================================
    # Discrete colour limits
    # ========================================================
    finite_abs = np.abs(
        best_corr[np.isfinite(best_corr)]
    )

    if finite_abs.size == 0:
        raise ValueError(
            "No valid correlation values to plot."
        )

    if colour_range is None:

        vmax = np.nanpercentile(
            finite_abs,
            auto_colour_percentile,
        )

        vmax = max(
            vmax,
            minimum_colour_range,
        )

        vmax = (
            np.ceil(vmax / colour_step)
            * colour_step
        )

        vmax = min(vmax, 1.0)

    else:
        vmax = abs(float(colour_range))

        vmax = (
            np.ceil(vmax / colour_step)
            * colour_step
        )

    boundaries = np.arange(
        -vmax,
        vmax + colour_step * 1.01,
        colour_step,
    )

    n_colours = len(boundaries) - 1

    cmap = plt.get_cmap(
        "RdBu_r",
        n_colours,
    )

    norm = BoundaryNorm(
        boundaries,
        cmap.N,
        clip=False,
    )

    # ========================================================
    # Plot
    # ========================================================
    fig, ax = plt.subplots(
        figsize=figsize
    )

    image = ax.imshow(
        np.ma.masked_invalid(best_corr),
        aspect="auto",
        interpolation="nearest",
        cmap=cmap,
        norm=norm,
    )

    # Predictor labels
    ax.set_xticks(
        np.arange(n_predictor)
    )

    ax.set_xticklabels(
        predictor_names,
        rotation=45,
        ha="right",
        fontsize=9,
    )

    # Region labels
    ax.set_yticks(
        np.arange(n_region)
    )

    ax.set_yticklabels(
        region_labels,
        fontsize=8.5,
    )

    # Cell annotation
    for ir in range(n_region):
        for ip in range(n_predictor):

            r = best_corr[ir, ip]

            if not np.isfinite(r):
                continue

            lead = best_lead[ir, ip]
            has_star = significant[ir, ip]

            # Contrast against background
            text_colour = (
                "white"
                if abs(r) >= 0.6 * vmax
                else "black"
            )

            if show_lead_numbers:
                label = str(lead)

                if has_star:
                    label += "*"

                ax.text(
                    ip,
                    ir,
                    label,
                    ha="center",
                    va="center",
                    fontsize=7.5,
                    fontweight=(
                        "bold"
                        if has_star
                        else "normal"
                    ),
                    color=text_colour,
                )

            elif has_star:
                ax.text(
                    ip,
                    ir,
                    "*",
                    ha="center",
                    va="center",
                    fontsize=10,
                    fontweight="bold",
                    color=text_colour,
                )

    # Cell boundaries
    ax.set_xticks(
        np.arange(-0.5, n_predictor, 1),
        minor=True,
    )

    ax.set_yticks(
        np.arange(-0.5, n_region, 1),
        minor=True,
    )

    ax.grid(
        which="minor",
        color="black",
        linewidth=0.35,
        alpha=0.25,
    )

    ax.tick_params(
        which="minor",
        bottom=False,
        left=False,
    )

    # ========================================================
    # Ocean group separators and labels
    # ========================================================
    ocean_display_default = {
        "SO": "Southern Ocean",
        "IO": "Indian Ocean",
        "PO": "Pacific Ocean",
        "NA": "North Atlantic",
        "SA": "South Atlantic",
        "Other": "Other",
    }

    if ocean_display is None:
        ocean_display = ocean_display_default
    else:
        ocean_display = {
            **ocean_display_default,
            **ocean_display,
        }

    group_start = 0
    groups = []

    for i in range(1, n_region + 1):

        if (
            i == n_region
            or region_oceans[i]
            != region_oceans[group_start]
        ):
            groups.append(
                (
                    group_start,
                    i - 1,
                    region_oceans[group_start],
                )
            )

            group_start = i

    for start, end, ocean in groups:

        if start > 0:
            ax.axhline(
                start - 0.5,
                color="black",
                linewidth=1.3,
            )

    transform = blended_transform_factory(
        ax.transAxes,
        ax.transData,
    )

    for start, end, ocean in groups:

        midpoint = 0.5 * (start + end)

        ax.text(
            -0.23,
            midpoint,
            ocean_display.get(ocean, ocean),
            transform=transform,
            ha="right",
            va="center",
            fontsize=9.5,
            fontweight="bold",
        )

    # ========================================================
    # Discrete colorbar
    # ========================================================
    colourbar = fig.colorbar(
        image,
        ax=ax,
        boundaries=boundaries,
        ticks=boundaries,
        spacing="uniform",
        extend="both",
        pad=0.015,
        fraction=0.036,
    )

    colourbar.set_label(
        "Correlation at strongest XRO lead",
        fontsize=10,
    )

    colourbar.ax.tick_params(
        labelsize=8
    )

    # ========================================================
    # Titles and notes
    # ========================================================
    if pp_rolling == 1:
        pp_description = "monthly PP"
    else:
        direction = (
            "centred"
            if rolling_center
            else "trailing"
        )

        pp_description = (
            f"{pp_rolling}-month {direction} mean PP"
        )

    if star_rule == "p":

        if significance_test == "permutation":
            significance_text = (
                f"* max-lag permutation "
                f"{'FDR q' if apply_fdr else 'p'} "
                f"< {alpha}"
            )
        else:
            significance_text = (
                f"* {'FDR q' if apply_fdr else 'p'} "
                f"< {alpha}"
            )

    elif star_rule == "corr":
        significance_text = (
            f"* |r| ≥ {sig_r}"
        )

    elif star_rule == "either":
        significance_text = (
            f"* p < {alpha} or |r| ≥ {sig_r}"
        )

    elif star_rule == "both":
        significance_text = (
            f"* p < {alpha} and |r| ≥ {sig_r}"
        )

    else:
        significance_text = "No significance markers"

    ax.set_title(
        "Strongest PP–XRO lead relationships\n"
        "Colour: signed correlation | "
        "Number: XRO lead month",
        fontsize=13,
        pad=12,
    )

    ax.set_xlabel(
        "XRO predictor",
        fontsize=10,
    )

    ax.set_ylabel(
        "PP region",
        fontsize=10,
    )

    fig.text(
        0.5,
        0.018,
        (
            f"{significance_text} | "
            f"Target: {pp_description} | "
            f"colour interval = {colour_step:.1f}"
        ),
        ha="center",
        va="bottom",
        fontsize=8.5,
    )

    fig.subplots_adjust(
        left=0.25,
        right=0.93,
        bottom=0.17,
        top=0.87,
    )

    if savepath is not None:
        fig.savefig(
            savepath,
            dpi=300,
            bbox_inches="tight",
        )

    plt.show()

    return result
```
